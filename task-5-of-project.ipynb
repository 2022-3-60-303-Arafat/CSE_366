```python
# tomato10_train_val_test_custom_cnn_256.py
# Single Kaggle cell: split tomato classes into train/val/test (70/10/20) and train custom CNN.
# Uses image size 256x256.

import os, random, shutil
from glob import glob
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import numpy as np
import matplotlib.pyplot as plt
import itertools
import tensorflow as tf
from tensorflow.keras import layers, models, callbacks, optimizers, regularizers
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# ---------------- USER / ENV SETTINGS ----------------
DATA_DIR = "/kaggle/input/plantdisease/PlantVillage"        # read-only on Kaggle
OUT_DIR = "/kaggle/working/subset_dataset"     # writable working dir
MODEL_PATH = "/kaggle/working/custom_cnn_tomato_best_256.h5"
IMG_SIZE = (256, 256)   # <-- changed to 256x256
BATCH_SIZE = 32
RANDOM_SEED = 1
VAL_RATIO = 0.10   # final desired proportion of total
TEST_RATIO = 0.20
TRAIN_RATIO = 0.70
EPOCHS = 20
INITIAL_LR = 1e-3
WEIGHT_DECAY = 1e-4
DROPOUT_RATE = 0.4
# ----------------------------------------------------

# exact tomato classes (saved preference)
tomato_classes = [
    'Tomato_Bacterial_spot',
    'Tomato_Early_blight',
    'Tomato_Late_blight',
    'Tomato_Leaf_Mold',
    'Tomato_Septoria_leaf_spot',
    'Tomato_Spider_mites_Two_spotted_spider_mite',
    'Tomato__Target_Spot',
    'Tomato__Tomato_YellowLeaf__Curl_Virus',
    'Tomato__Tomato_mosaic_virus',
    'Tomato_healthy'
]

# reproducibility
random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
tf.random.set_seed(RANDOM_SEED)

Path(OUT_DIR).mkdir(parents=True, exist_ok=True)

# 0) verify classes exist in DATA_DIR
available = [d for d in sorted(os.listdir(DATA_DIR)) if os.path.isdir(os.path.join(DATA_DIR, d))]
missing = [c for c in tomato_classes if c not in available]
if missing:
    print("Warning: the following requested classes were NOT found under", DATA_DIR)
    for m in missing:
        print("  -", m)
selected = [c for c in tomato_classes if c in available]
if len(selected) == 0:
    raise SystemExit("No requested tomato classes exist under DATA_DIR. Check names.")
print("Using classes ({}):".format(len(selected)))
for c in selected:
    print(" -", c)

# 1) Create train/val/test splits per class and copy files
def prepare_train_val_test(data_dir, out_dir, classes, train_ratio=0.7, val_ratio=0.1, test_ratio=0.2, seed=42):
    assert abs(train_ratio + val_ratio + test_ratio - 1.0) < 1e-6
    out_train = os.path.join(out_dir, "train")
    out_val = os.path.join(out_dir, "val")
    out_test = os.path.join(out_dir, "test")
    Path(out_train).mkdir(parents=True, exist_ok=True)
    Path(out_val).mkdir(parents=True, exist_ok=True)
    Path(out_test).mkdir(parents=True, exist_ok=True)

    for cls in classes:
        src_dir = os.path.join(data_dir, cls)
        images = sorted(glob(os.path.join(src_dir, "*")))
        images = [p for p in images if p.lower().endswith(('.jpg','.jpeg','.png','.bmp'))]
        n = len(images)
        if n == 0:
            print(f"Skipping {cls} — no images found.")
            continue

        # Step 1: split off test (test_ratio)
        test_size = test_ratio
        rest_size = 1.0 - test_size
        images_trainval, images_test = train_test_split(images, test_size=test_size, random_state=seed, shuffle=True)

        # Step 2: from the remaining, compute val fraction (val_ratio / rest_size)
        if rest_size <= 0:
            images_train, images_val = [], []
        else:
            val_frac_of_rest = val_ratio / rest_size  # e.g., 0.1 / 0.8 = 0.125
            images_train, images_val = train_test_split(images_trainval, test_size=val_frac_of_rest, random_state=seed, shuffle=True)

        # Copy files to corresponding folders
        dst_train_cls = os.path.join(out_train, cls); Path(dst_train_cls).mkdir(parents=True, exist_ok=True)
        dst_val_cls = os.path.join(out_val, cls); Path(dst_val_cls).mkdir(parents=True, exist_ok=True)
        dst_test_cls = os.path.join(out_test, cls); Path(dst_test_cls).mkdir(parents=True, exist_ok=True)

        for f in images_train:
            shutil.copy(f, dst_train_cls)
        for f in images_val:
            shutil.copy(f, dst_val_cls)
        for f in images_test:
            shutil.copy(f, dst_test_cls)

        print(f"Class {cls}: total={n}, train={len(images_train)}, val={len(images_val)}, test={len(images_test)}")
    return out_train, out_val, out_test

train_folder, val_folder, test_folder = prepare_train_val_test(DATA_DIR, OUT_DIR, selected, TRAIN_RATIO, VAL_RATIO, TEST_RATIO, RANDOM_SEED)

# quick totals
def folder_count(folder):
    total = 0
    for root, dirs, files in os.walk(folder):
        total += len([f for f in files if f.lower().endswith(('.jpg','.jpeg','.png','.bmp'))])
    return total

print("Totals -> train:", folder_count(train_folder), " val:", folder_count(val_folder), " test:", folder_count(test_folder))

# 2) Data augmentation & generators (with 256x256 target size)
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.12,
    height_shift_range=0.12,
    shear_range=0.08,
    zoom_range=0.12,
    horizontal_flip=True,
    brightness_range=(0.8, 1.2),
    fill_mode='nearest'
)

val_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    train_folder,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=True
)
val_gen = val_datagen.flow_from_directory(
    val_folder,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)
test_gen = test_datagen.flow_from_directory(
    test_folder,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

num_classes = len(train_gen.class_indices)
print("Num classes (subset):", num_classes)
print("Class indices:", train_gen.class_indices)

# 3) Optional class weights for imbalance
from sklearn.utils.class_weight import compute_class_weight
y_train = train_gen.classes
class_weights = compute_class_weight(class_weight='balanced', classes=np.unique(y_train), y=y_train)
class_weight_dict = {i: float(w) for i, w in enumerate(class_weights)}
print("Class weights:", class_weight_dict)

# 4) Build custom CNN (input adjusted for 256x256)
def build_custom_cnn(input_shape=(256,256,3), num_classes=10, dropout_rate=0.4, weight_decay=0.0):
    inp = layers.Input(shape=input_shape)

    # Block 1
    x = layers.Conv2D(32, (3,3), padding='same', kernel_initializer='he_normal')(inp)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Conv2D(32, (3,3), padding='same', kernel_initializer='he_normal')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.MaxPooling2D((2,2))(x)
    x = layers.Dropout(0.15)(x)

    # Block 2
    x = layers.Conv2D(64, (3,3), padding='same', kernel_initializer='he_normal')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Conv2D(64, (3,3), padding='same', kernel_initializer='he_normal')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.MaxPooling2D((2,2))(x)
    x = layers.Dropout(0.2)(x)

    # Block 3
    x = layers.Conv2D(128, (3,3), padding='same', kernel_initializer='he_normal')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Conv2D(128, (3,3), padding='same', kernel_initializer='he_normal')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.MaxPooling2D((2,2))(x)
    x = layers.Dropout(0.25)(x)

    # Block 4
    x = layers.Conv2D(256, (3,3), padding='same', kernel_initializer='he_normal')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Conv2D(256, (3,3), padding='same', kernel_initializer='he_normal')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.MaxPooling2D((2,2))(x)
    x = layers.Dropout(0.3)(x)

    # Global pooling + FC head
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(512, kernel_regularizer=regularizers.l2(weight_decay))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Dropout(dropout_rate)(x)

    x = layers.Dense(256, kernel_regularizer=regularizers.l2(weight_decay))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Dropout(dropout_rate)(x)

    out = layers.Dense(num_classes, activation='softmax')(x)

    model = models.Model(inputs=inp, outputs=out)
    return model

model = build_custom_cnn(input_shape=IMG_SIZE + (3,), num_classes=num_classes, dropout_rate=DROPOUT_RATE, weight_decay=WEIGHT_DECAY)
model.summary()

# 5) Compile
opt = optimizers.Adam(learning_rate=INITIAL_LR)
model.compile(optimizer=opt, loss='categorical_crossentropy', metrics=['accuracy'])

# 6) Callbacks
ckpt = callbacks.ModelCheckpoint(MODEL_PATH, monitor='val_accuracy', save_best_only=True, verbose=1)
rlr = callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, verbose=1)
es = callbacks.EarlyStopping(monitor='val_loss', patience=7, restore_best_weights=True, verbose=1)

# ---------------- REPLACEMENT: Train, evaluate, metrics, ROC, timings, paired t-test ----------------

from sklearn.metrics import (classification_report, confusion_matrix,
                             precision_recall_fscore_support, accuracy_score,
                             roc_auc_score, roc_curve, auc)
from sklearn.preprocessing import label_binarize
from scipy.stats import ttest_rel
import time

# 7) Train (monitor validation set) — measure training time
steps_per_epoch = int(np.ceil(train_gen.samples / train_gen.batch_size))
validation_steps = int(np.ceil(val_gen.samples / val_gen.batch_size))

print("Starting training...")
train_start = time.time()
history = model.fit(
    train_gen,
    epochs=EPOCHS,
    steps_per_epoch=steps_per_epoch,
    validation_data=val_gen,
    validation_steps=validation_steps,
    callbacks=[ckpt, rlr, es],
    class_weight=class_weight_dict
)
train_end = time.time()
train_time = train_end - train_start
print(f"Training finished. Training time: {train_time:.2f} seconds ({train_time/60:.2f} minutes)")

# 8) Evaluate on TEST set (load best model) and measure inference time
model.load_weights(MODEL_PATH)
test_steps = int(np.ceil(test_gen.samples / test_gen.batch_size))

print("Running inference on test set...")
test_start = time.time()
y_pred_probas = model.predict(test_gen, steps=test_steps, verbose=1)
test_end = time.time()
test_time = test_end - test_start
print(f"Inference finished. Test (inference) time for full test set: {test_time:.4f} seconds")

# predicted classes
if y_pred_probas.ndim == 1 or (y_pred_probas.ndim == 2 and y_pred_probas.shape[1] == 1):
    # binary-ish output -> threshold
    y_pred_classes = (y_pred_probas.ravel() > 0.5).astype(int)
else:
    y_pred_classes = np.argmax(y_pred_probas, axis=1)

y_true = np.array(test_gen.classes)
class_names = list(test_gen.class_indices.keys())
labels = np.unique(y_true)
num_classes = len(class_names)

print("\n=== TEST SET RESULTS ===\n")
print("Test samples:", test_gen.samples)
print("Num classes:", num_classes)
print("Class indices:", test_gen.class_indices)

# Overall accuracy
overall_acc = accuracy_score(y_true, y_pred_classes)
print(f"\nOverall Accuracy: {overall_acc:.4f}")

# Per-class accuracy
per_class_acc = {}
for lbl in labels:
    idx = np.where(y_true == lbl)[0]
    if len(idx) == 0:
        per_class_acc[lbl] = np.nan
    else:
        per_class_acc[lbl] = np.mean(y_pred_classes[idx] == y_true[idx])
print("\nPer-class accuracy (label: accuracy):")
for lbl in labels:
    cname = class_names[lbl] if lbl < len(class_names) else str(lbl)
    print(f"  {lbl} ({cname}): {per_class_acc[lbl]:.4f}")

# Precision/Recall/F1 per-class and macro/micro
prec, rec, f1, sup = precision_recall_fscore_support(y_true, y_pred_classes, average=None, zero_division=0)
pr_df = pd.DataFrame({
    'class_index': labels,
    'class_name': [class_names[i] for i in labels],
    'precision': prec,
    'recall': rec,
    'f1-score': f1,
    'support': sup
})
print("\nPer-class precision / recall / f1 / support:")
display(pr_df)

macro_prec, macro_rec, macro_f1, _ = precision_recall_fscore_support(y_true, y_pred_classes, average='macro', zero_division=0)
micro_prec, micro_rec, micro_f1, _ = precision_recall_fscore_support(y_true, y_pred_classes, average='micro', zero_division=0)
print(f"\nMacro Precision: {macro_prec:.4f}, Macro Recall: {macro_rec:.4f}, Macro F1: {macro_f1:.4f}")
print(f"Micro Precision: {micro_prec:.4f}, Micro Recall: {micro_rec:.4f}, Micro F1: {micro_f1:.4f}")

# Confusion matrix & classification report
cm = confusion_matrix(y_true, y_pred_classes, labels=labels)
print("\nConfusion Matrix:")
display(pd.DataFrame(cm, index=[f"true_{class_names[i]}" for i in labels],
                     columns=[f"pred_{class_names[i]}" for i in labels]))

print("\nClassification Report:")
print(classification_report(y_true, y_pred_classes, target_names=[class_names[i] for i in labels], zero_division=0))

# ---------- AUC & ROC (one-vs-rest for multiclass) ----------
try:
    if num_classes == 2:
        # binary
        if y_pred_probas.ndim > 1 and y_pred_probas.shape[1] > 1:
            pos_prob = y_pred_probas[:, 1]
        else:
            pos_prob = y_pred_probas.ravel()
        auc_score = roc_auc_score(y_true, pos_prob)
        fpr, tpr, _ = roc_curve(y_true, pos_prob)
        print(f"\nROC AUC (binary): {auc_score:.4f}")
        plt.figure()
        plt.plot(fpr, tpr, lw=2, label=f'AUC = {auc_score:.3f}')
        plt.plot([0,1],[0,1], linestyle='--')
        plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate'); plt.title('ROC Curve (binary)')
        plt.legend(loc='lower right'); plt.show()
    else:
        # multiclass: one-vs-rest
        y_true_bin = label_binarize(y_true, classes=labels)  # shape (n_samples, n_classes)
        if y_pred_probas.shape[1] != len(labels):
            raise ValueError(f"y_pred_probas shape {y_pred_probas.shape} doesn't match number of classes {len(labels)}")
        fpr = dict(); tpr = dict(); roc_auc = dict()
        for i, lab in enumerate(labels):
            fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], y_pred_probas[:, i])
            roc_auc[i] = auc(fpr[i], tpr[i])
        # micro-average
        fpr["micro"], tpr["micro"], _ = roc_curve(y_true_bin.ravel(), y_pred_probas.ravel())
        roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])
        # macro-average
        all_fpr = np.unique(np.concatenate([fpr[i] for i in range(len(labels))]))
        mean_tpr = np.zeros_like(all_fpr)
        for i in range(len(labels)):
            mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])
        mean_tpr /= len(labels)
        fpr["macro"] = all_fpr
        tpr["macro"] = mean_tpr
        roc_auc["macro"] = auc(fpr["macro"], tpr["macro"])

        print("\nAUC per class:")
        for i, lab in enumerate(labels):
            print(f"  {lab} ({class_names[lab]}): AUC = {roc_auc[i]:.4f}")
        print(f"micro-average AUC: {roc_auc['micro']:.4f}")
        print(f"macro-average AUC: {roc_auc['macro']:.4f}")

        # Plot ROC curves
        plt.figure(figsize=(9,7))
        plt.plot(fpr["micro"], tpr["micro"], label=f'micro-average ROC (AUC = {roc_auc["micro"]:.3f})', linewidth=2)
        plt.plot(fpr["macro"], tpr["macro"], label=f'macro-average ROC (AUC = {roc_auc["macro"]:.3f})', linewidth=2, linestyle='--')
        for i, lab in enumerate(labels):
            plt.plot(fpr[i], tpr[i], lw=1, label=f'{class_names[lab]} (AUC = {roc_auc[i]:.3f})')
        plt.plot([0,1],[0,1],'k--', lw=0.8)
        plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate'); plt.title('Multi-class ROC curves')
        plt.legend(loc='lower right', fontsize='small'); plt.show()
except Exception as e:
    print("Could not compute ROC/AUC:", e)

# 9) Plot confusion matrix and training curves (reuse your plotting function)
def plot_confusion_matrix(cm, class_names):
    plt.figure(figsize=(10,10))
    plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
    plt.title("Confusion matrix (test set)")
    plt.colorbar()
    ticks = np.arange(len(class_names))
    plt.xticks(ticks, class_names, rotation=90)
    plt.yticks(ticks, class_names)
    fmt = 'd'
    thresh = cm.max() / 2. if cm.size>0 else 0
    for i, j in itertools.product(range(cm.shape[0]), range(cm.shape[1])):
        plt.text(j, i, format(cm[i, j], fmt),
                 horizontalalignment="center",
                 color="white" if cm[i, j] > thresh else "black")
    plt.tight_layout()
    plt.show()

plot_confusion_matrix(cm, [class_names[i] for i in labels])

# Plot train/val curves
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.plot(history.history.get('loss', []), label='train_loss')
plt.plot(history.history.get('val_loss', []), label='val_loss')
plt.legend(); plt.title('Loss')
plt.subplot(1,2,2)
plt.plot(history.history.get('accuracy', []), label='train_acc')
plt.plot(history.history.get('val_accuracy', []), label='val_acc')
plt.legend(); plt.title('Accuracy')
plt.show()

print("Best model saved to:", MODEL_PATH)
print("Subset data at:", OUT_DIR)
print(f"Training time (s): {train_time:.2f}, Test inference time (s): {test_time:.4f}")

# --------------- Paired t-test (if you have two result vectors to compare) ---------------
# To run a paired t-test you must provide two 1-D arrays of the same length:
# Example usage (uncomment and set these arrays from repeated runs / cross-val folds):
# scores_model1 = np.array([0.91,0.93,0.90,0.92,0.94])
# scores_model2 = np.array([0.89,0.92,0.88,0.90,0.91])
scores_model1 = globals().get('scores_model1', None)
scores_model2 = globals().get('scores_model2', None)

if scores_model1 is not None and scores_model2 is not None:
    scores_model1 = np.asarray(scores_model1)
    scores_model2 = np.asarray(scores_model2)
    if scores_model1.shape != scores_model2.shape:
        print("scores_model1 and scores_model2 must have the same shape for paired t-test.")
    else:
        t_stat, p_val = ttest_rel(scores_model1, scores_model2)
        print("\nPaired t-test comparing two sets of scores:")
        print(f"  t-statistic = {t_stat:.4f}, p-value = {p_val:.4e}")
        alpha = 0.05
        if p_val < alpha:
            print(f"  => Reject null hypothesis at alpha={alpha}: significant difference.")
        else:
            print(f"  => Fail to reject null at alpha={alpha}: no significant difference detected.")
else:
    print("\nPaired t-test not run: provide `scores_model1` and `scores_model2` arrays in the notebook to execute it.")
    print("Example: scores_model1 = [0.91,0.93,...]; scores_model2 = [0.89,0.92,...]  (same length)")

# Save summary (optional)
summary = {
    'overall_accuracy': overall_acc,
    'macro_precision': macro_prec,
    'macro_recall': macro_rec,
    'macro_f1': macro_f1,
    'train_time_seconds': train_time,
    'test_time_seconds': test_time
}
print("\nSummary:")
display(pd.DataFrame([summary]))

 
```

```python
# ======================================================
#  Task 5.1 – Robust Grad-CAM setup (use model_gc)
# ======================================================
import cv2
from tensorflow.keras.preprocessing import image
from tensorflow.keras.layers import Conv2D

print("\n[Grad-CAM] Using model_gc for explainability.")

# 1) Find LAST Conv2D layer by type, not by shape
def find_last_conv_layer(model):
    for layer in reversed(model.layers):
        if isinstance(layer, Conv2D):
            return layer.name
    raise ValueError("No Conv2D layer found in model. Check architecture.")

last_conv_layer_name = find_last_conv_layer(model_gc)
print("Last Conv2D layer selected for Grad-CAM:", last_conv_layer_name)

# 2) Preprocessing (same as training)
def load_and_prep(path, target_size=IMG_SIZE):
    img = image.load_img(path, target_size=target_size)
    arr = image.img_to_array(img).astype("float32") / 255.0
    return np.expand_dims(arr, 0), np.array(img, dtype=np.uint8)

# 3) Grad-CAM core
def make_gradcam_heatmap(img_array, model, last_conv_name, pred_index=None):
    grad_model = tf.keras.models.Model(
        [model.inputs],
        [model.get_layer(last_conv_name).output, model.output]
    )

    with tf.GradientTape() as tape:
        conv_output, preds = grad_model(img_array)
        if pred_index is None:
            pred_index = tf.argmax(preds[0])
        class_score = preds[:, pred_index]

    grads = tape.gradient(class_score, conv_output)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    conv_output = conv_output[0]

    heatmap = conv_output @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)
    heatmap = tf.maximum(heatmap, 0) / (tf.reduce_max(heatmap) + 1e-8)

    return heatmap.numpy()

# 4) Overlay heatmap
def overlay_heatmap(img, heatmap, alpha=0.4):
    heatmap = cv2.resize(heatmap, (img.shape[1], img.shape[0]))
    heatmap = np.uint8(255 * heatmap)
    heatmap = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)
    return cv2.addWeighted(img, 1-alpha, heatmap, alpha, 0)
```

```python
# ======================================================
#  Task 5.1 – Run Grad-CAM on test samples
# ======================================================

# Build list of paths from your existing split
test_image_paths = []
class_indices = test_gen.class_indices
inv_class_indices = {v: k for k, v in class_indices.items()}

for cls_name in class_indices.keys():
    cls_dir = os.path.join(test_folder, cls_name)
    test_image_paths.extend(
        glob(os.path.join(cls_dir, "*.jpg")) +
        glob(os.path.join(cls_dir, "*.jpeg")) +
        glob(os.path.join(cls_dir, "*.png"))
    )

print("Total test images found:", len(test_image_paths))

# True labels
true_labels_dict = {
    p: class_indices[os.path.basename(os.path.dirname(p))]
    for p in test_image_paths
}

# Grad-CAM on a random subset
results = []   # (path, true_idx, pred_idx, correct, overlay_img)
sample_paths = random.sample(test_image_paths, min(60, len(test_image_paths)))

print("\nComputing Grad-CAM heatmaps...")
for p in sample_paths:
    x, orig = load_and_prep(p)
    preds = model_gc.predict(x, verbose=0)
    pred_idx = int(np.argmax(preds[0]))
    true_idx = true_labels_dict[p]
    correct = (pred_idx == true_idx)

    heatmap = make_gradcam_heatmap(x, model_gc, last_conv_layer_name, pred_index=pred_idx)
    overlay = overlay_heatmap(orig, heatmap)

    results.append((p, true_idx, pred_idx, correct, overlay))

correct_samples = [r for r in results if r[3]]
incorrect_samples = [r for r in results if not r[3]]

print("Grad-CAM samples -> Correct:", len(correct_samples), " Incorrect:", len(incorrect_samples))
```
