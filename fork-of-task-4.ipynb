```python


# ---------------------------
# Imports & device detection
# ---------------------------
import os, sys, time, json, random
from pathlib import Path
from typing import List, Dict, Tuple, Any

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, random_split
from torchvision import datasets, transforms
from torchvision.datasets.folder import default_loader

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, roc_auc_score, roc_curve, classification_report
)

# ---------------------------
# User-specified tomato classes and input path
# ---------------------------
TOMATO_CLASSES = [
    'Tomato_Bacterial_spot',
    'Tomato_Early_blight',
    'Tomato_Late_blight',
    'Tomato_Leaf_Mold',
    'Tomato_Septoria_leaf_spot',
    'Tomato_Spider_mites_Two_spotted_spider_mite',
    'Tomato__Target_Spot',
    'Tomato__Tomato_YellowLeaf__Curl_Virus',
    'Tomato__Tomato_mosaic_virus',
    'Tomato_healthy'
]

PLANTVILLAGE_ROOT = Path("/kaggle/input/plantdisease/PlantVillage")
if not PLANTVILLAGE_ROOT.exists():
    raise FileNotFoundError(f"Dataset path not found: {PLANTVILLAGE_ROOT}. Attach dataset to notebook Input.")

# ---------------------------
# Device and reproducibility
# ---------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False
GENERATOR = torch.Generator().manual_seed(SEED)

# ---------------------------
# Hyperparameters (user-editable)
# ---------------------------
DATA_DIR = str(PLANTVILLAGE_ROOT)
DATASET_NAME = "PlantVillage_tomato_subset"
IMG_SIZE = 224
BATCH_SIZE = 32
LR = 1e-4
EPOCHS = 30                                  # requested
EARLY_STOPPING_PATIENCE = 10                 # requested
ATTENTION_MODULE_NAME = "SE"
CHECKPOINT_PATH = f"/kaggle/working/task4_attention_{ATTENTION_MODULE_NAME}_{DATASET_NAME}.pth"
NUM_WORKERS = 2

# ---------------------------
# Transforms and dataset filtering (tomato-only)
# ---------------------------
train_transforms = transforms.Compose([
    transforms.RandomHorizontalFlip(0.5),
    transforms.RandomRotation(10),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])
])
val_test_transforms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])
])

full_dataset_all = datasets.ImageFolder(root=DATA_DIR, transform=train_transforms)
all_classes = full_dataset_all.classes
print("Total classes found:", len(all_classes))

missing = [c for c in TOMATO_CLASSES if c not in all_classes]
if missing:
    raise ValueError(f"Missing tomato class directories under {DATA_DIR}: {missing}")

# collect sample indices for only tomato classes
allowed_class_indices = {full_dataset_all.class_to_idx[c] for c in TOMATO_CLASSES}
selected_sample_indices: List[int] = [
    i for i, (_, target) in enumerate(full_dataset_all.samples) if target in allowed_class_indices
]
if len(selected_sample_indices) == 0:
    raise RuntimeError("No samples found for tomato classes — check dataset structure.")

# remap original class indices -> 0..9 according to TOMATO_CLASSES order
orig_class_to_idx = full_dataset_all.class_to_idx
tomato_orig_to_new = {orig_class_to_idx[name]: i for i, name in enumerate(TOMATO_CLASSES)}
new_class_names = TOMATO_CLASSES.copy()
NUM_CLASSES = len(new_class_names)
print("Using tomato classes (NUM_CLASSES):", NUM_CLASSES)
print("Total tomato images selected:", len(selected_sample_indices))

# wrapper to present only selected samples with remapped labels and separate transforms
class FilteredImageFolder(torch.utils.data.Dataset):
    def __init__(self, image_folder_dataset, selected_indices: List[int], orig_to_new_label: Dict[int,int], transform=None, loader=default_loader):
        self.image_folder = image_folder_dataset
        self.selected_indices = selected_indices
        self.orig_to_new = orig_to_new_label
        self.transform = transform if transform is not None else image_folder_dataset.transform
        self.loader = loader
        self.samples = [self.image_folder.samples[i] for i in selected_indices]
        self._len = len(self.samples)
    def __len__(self): return self._len
    def __getitem__(self, idx):
        path, orig_label = self.samples[idx]
        img = self.loader(path)
        if self.transform: img = self.transform(img)
        return img, self.orig_to_new[orig_label]

# instantiate filtered_full with train transforms initially
filtered_full = FilteredImageFolder(full_dataset_all, selected_sample_indices, tomato_orig_to_new, transform=train_transforms)

# ---------------------------
# Deterministic 70/20/10 split
# ---------------------------
total = len(filtered_full)
n_train = int(0.70 * total)
n_val = int(0.20 * total)
n_test = total - n_train - n_val
if n_test <= 0:
    raise ValueError("Not enough samples for 10% test split after filtering.")

train_ds, val_ds, test_ds = random_split(filtered_full, [n_train, n_val, n_test], generator=GENERATOR)
train_indices = train_ds.indices if hasattr(train_ds, "indices") else list(range(0, n_train))
val_indices   = val_ds.indices   if hasattr(val_ds, "indices") else list(range(n_train, n_train+n_val))
test_indices  = test_ds.indices  if hasattr(test_ds, "indices") else list(range(n_train+n_val, total))

train_dataset = FilteredImageFolder(full_dataset_all, [selected_sample_indices[i] for i in train_indices], tomato_orig_to_new, transform=train_transforms)
val_dataset   = FilteredImageFolder(full_dataset_all, [selected_sample_indices[i] for i in val_indices], tomato_orig_to_new, transform=val_test_transforms)
test_dataset  = FilteredImageFolder(full_dataset_all, [selected_sample_indices[i] for i in test_indices], tomato_orig_to_new, transform=val_test_transforms)

print("Split sizes (train/val/test):", len(train_dataset), len(val_dataset), len(test_dataset))

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available())
val_loader   = DataLoader(val_dataset,   batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available())
test_loader  = DataLoader(test_dataset,  batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available())

# ---------------------------
# SE attention module (pure PyTorch)
# ---------------------------
class SEBlock(nn.Module):
    def __init__(self, channels:int, reduction:int=16):
        super().__init__()
        reduced = max(1, channels//reduction)
        self.fc1 = nn.Linear(channels, reduced)
        self.fc2 = nn.Linear(reduced, channels)
    def forward(self, x):
        b,c,_,_ = x.size()
        y = x.view(b,c,-1).mean(dim=2)             # GAP
        y = self.fc1(y)
        y = F.relu(y, inplace=True)
        y = self.fc2(y)
        y = torch.sigmoid(y).view(b,c,1,1)
        return x * y

# ---------------------------
# Task-3 CNN reference + Attention CNN (SE inserted after last conv block)
# ---------------------------
class Task3CNN(nn.Module):
    def __init__(self, num_classes:int=10):
        super().__init__()
        def conv_block(in_ch,out_ch):
            return nn.Sequential(
                nn.Conv2d(in_ch,out_ch,kernel_size=3,padding=1,bias=False),
                nn.BatchNorm2d(out_ch),
                nn.ReLU(inplace=True),
                nn.MaxPool2d(2)
            )
        self.conv1 = conv_block(3,32)
        self.conv2 = conv_block(32,64)
        self.conv3 = conv_block(64,128)
        self.conv4 = conv_block(128,256)   # last conv block
        self.global_pool = nn.AdaptiveAvgPool2d(1)
        self.classifier = nn.Linear(256, num_classes)
    def forward(self,x):
        x = self.conv1(x); x = self.conv2(x); x = self.conv3(x); x = self.conv4(x)
        x = self.global_pool(x)
        x = x.view(x.size(0), -1)
        x = self.classifier(x)
        return x

class AttentionCNN(Task3CNN):
    def __init__(self, num_classes:int=10):
        super().__init__(num_classes=num_classes)
        self.se = SEBlock(channels=256, reduction=16)
    def forward(self,x):
        x = self.conv1(x); x = self.conv2(x); x = self.conv3(x); x = self.conv4(x)
        x = self.se(x)            # attention injected here (before GAP)
        x = self.global_pool(x)
        x = x.view(x.size(0), -1)
        x = self.classifier(x)
        return x

# parameter check (warn if >5% increase)
def count_parameters(m): return sum(p.numel() for p in m.parameters() if p.requires_grad)
base_model = Task3CNN(num_classes=NUM_CLASSES)
att_model  = AttentionCNN(num_classes=NUM_CLASSES)
base_params = count_parameters(base_model)
att_params  = count_parameters(att_model)
pct_increase = (att_params - base_params) / base_params * 100.0
print(f"Params: base={base_params:,}, attention={att_params:,}, increase={pct_increase:.3f}%")
if pct_increase > 5.0:
    print("WARNING: parameter increase > 5% — consider larger SE reduction.")

# ---------------------------
# Training / evaluation utilities
# ---------------------------
def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()
    running_loss = 0.0
    all_preds, all_targets = [], []
    for images, targets in loader:
        images = images.to(device); targets = targets.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * images.size(0)
        preds = outputs.argmax(dim=1).detach().cpu().numpy()
        all_preds.append(preds); all_targets.append(targets.detach().cpu().numpy())
    avg_loss = running_loss / len(loader.dataset)
    if len(all_preds):
        all_preds = np.concatenate(all_preds); all_targets = np.concatenate(all_targets)
        f1 = f1_score(all_targets, all_preds, average='macro', zero_division=0)
    else:
        f1 = 0.0
    return avg_loss, f1

def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    all_preds, all_targets, all_probs = [], [], []
    with torch.no_grad():
        for images, targets in loader:
            images = images.to(device); targets = targets.to(device)
            outputs = model(images)
            loss = criterion(outputs, targets)
            running_loss += loss.item() * images.size(0)
            probs = F.softmax(outputs, dim=1).detach().cpu().numpy()
            preds = probs.argmax(axis=1)
            all_probs.append(probs); all_preds.append(preds); all_targets.append(targets.detach().cpu().numpy())
    avg_loss = running_loss / len(loader.dataset)
    all_preds = np.concatenate(all_preds) if all_preds else np.array([])
    all_targets = np.concatenate(all_targets) if all_targets else np.array([])
    all_probs = np.vstack(all_probs) if all_probs else np.zeros((len(all_targets), NUM_CLASSES))
    f1 = f1_score(all_targets, all_preds, average='macro', zero_division=0) if len(all_preds) else 0.0
    return avg_loss, f1, all_targets, all_preds, all_probs

def fit(model, train_loader, val_loader, device, epochs=30, patience=10, lr=1e-4):
    model.to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    criterion = nn.CrossEntropyLoss()
    best_val_f1 = -1.0
    best_state = None
    epochs_no_improve = 0
    history = {'train_loss':[],'train_f1':[],'val_loss':[],'val_f1':[]}
    for epoch in range(1, epochs+1):
        train_loss, train_f1 = train_one_epoch(model, train_loader, optimizer, criterion, device)
        val_loss, val_f1, *_ = evaluate(model, val_loader, criterion, device)
        history['train_loss'].append(train_loss); history['train_f1'].append(train_f1)
        history['val_loss'].append(val_loss); history['val_f1'].append(val_f1)
        tqdm.write(f"Epoch {epoch:02d} | train_loss {train_loss:.4f} train_f1 {train_f1:.4f} | val_loss {val_loss:.4f} val_f1 {val_f1:.4f}")
        if val_f1 > best_val_f1 + 1e-6:
            best_val_f1 = val_f1
            best_state = {k:v.cpu() for k,v in model.state_dict().items()}
            epochs_no_improve = 0
        else:
            epochs_no_improve += 1
        if epochs_no_improve >= patience:
            tqdm.write(f"Early stopping: no improvement in val_f1 for {patience} epochs.")
            break
    if best_state is not None:
        model.load_state_dict(best_state)
    return model, history

# ---------------------------
# Train (measure full training wall-clock)
# ---------------------------
model = AttentionCNN(num_classes=NUM_CLASSES).to(device)
print("Starting training: epochs (max) =", EPOCHS, ", patience =", EARLY_STOPPING_PATIENCE)
start_time = time.time()
if torch.cuda.is_available(): torch.cuda.synchronize()
model, history = fit(model, train_loader, val_loader, device, epochs=EPOCHS, patience=EARLY_STOPPING_PATIENCE, lr=LR)
if torch.cuda.is_available(): torch.cuda.synchronize()
training_time_secs = time.time() - start_time
print(f"Training wall-clock time (seconds): {training_time_secs:.3f}")

# ---------------------------
# Plot train/val loss & F1 curves
# ---------------------------
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.plot(history['train_loss'], label='train_loss')
plt.plot(history['val_loss'], label='val_loss')
plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.title('Loss'); plt.legend(); plt.grid(alpha=0.3)
plt.subplot(1,2,2)
plt.plot(history['train_f1'], label='train_f1')
plt.plot(history['val_f1'], label='val_f1')
plt.xlabel('Epoch'); plt.ylabel('F1-macro'); plt.title('F1 (macro)'); plt.legend(); plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# ---------------------------
# Evaluate on test set & metrics
# ---------------------------
criterion = nn.CrossEntropyLoss()
test_loss, test_f1, y_true, y_pred, y_prob = evaluate(model, test_loader, criterion, device)

accuracy = accuracy_score(y_true, y_pred)
precision_macro = precision_score(y_true, y_pred, average='macro', zero_division=0)
recall_macro = recall_score(y_true, y_pred, average='macro', zero_division=0)
f1_macro = f1_score(y_true, y_pred, average='macro', zero_division=0)

cm = confusion_matrix(y_true, y_pred)
per_class_acc = {}
for i, cname in enumerate(new_class_names):
    true_count = (y_true == i).sum()
    correct = cm[i, i]
    per_class_acc[cname] = (correct / true_count) if true_count>0 else 0.0

try:
    y_true_onehot = np.zeros((len(y_true), NUM_CLASSES))
    y_true_onehot[np.arange(len(y_true)), y_true] = 1
    auc_ovr = roc_auc_score(y_true_onehot, y_prob, average='macro', multi_class='ovr')
except Exception:
    auc_ovr = None

metrics_table = pd.DataFrame({
    'Metric':['Accuracy','Precision (macro)','Recall (macro)','F1 (macro)','AUC (OvR)'],
    'Value':[f"{accuracy:.4f}", f"{precision_macro:.4f}", f"{recall_macro:.4f}", f"{f1_macro:.4f}", f"{auc_ovr:.4f}" if auc_ovr is not None else 'N/A']
})
print("\n=== Test Metrics ===")
display(metrics_table)

print("\nPer-class accuracy:")
for k,v in per_class_acc.items(): print(f"{k}: {v:.3f}")

print("\nClassification report (digits=3):")
print(classification_report(y_true, y_pred, target_names=new_class_names, digits=3, zero_division=0))

print("\nConfusion matrix:")
display(pd.DataFrame(cm, index=new_class_names, columns=new_class_names))

# ---------------------------
# ROC curves (one per class)
# ---------------------------
plt.figure(figsize=(10,8))
for i, cname in enumerate(new_class_names):
    try:
        fpr, tpr, _ = roc_curve(y_true==i, y_prob[:, i])
        auc_score = roc_auc_score((y_true==i).astype(int), y_prob[:, i])
        plt.plot(fpr, tpr, label=f"{cname} (AUC={auc_score:.3f})")
    except Exception:
        continue
plt.plot([0,1],[0,1],'--',linewidth=0.5)
plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate')
plt.title('ROC curves — Test set'); plt.legend(loc='lower right', fontsize='small'); plt.grid(alpha=0.3)
plt.show()

# ---------------------------
# Inference time (avg over 3 runs)
# ---------------------------
def measure_inference_time(model, loader, device, runs=3):
    model.eval()
    times=[]
    with torch.no_grad():
        for _ in range(runs):
            start=time.time()
            for images, _ in loader:
                images = images.to(device)
                _ = model(images)
            if torch.cuda.is_available(): torch.cuda.synchronize()
            times.append(time.time()-start)
    return np.mean(times), np.std(times)

inf_mean, inf_std = measure_inference_time(model, test_loader, device, runs=3)
print(f"\nInference full-test time avg (3 runs): {inf_mean:.3f}s ± {inf_std:.3f}s")
print(f"Training wall-clock time for this run: {training_time_secs:.3f}s")

# ---------------------------
# Comparison table: append Task-4 row to existing summary if present
# ---------------------------
summary_path = '/kaggle/working/task2_3_summary.csv'
if Path(summary_path).exists():
    summary_df = pd.read_csv(summary_path)
else:
    summary_df = pd.DataFrame([
       
    {'Task': 'Task-2', 'Model': 'VGG16',          'Accuracy': '0.92675', 'F1_macro': '0.91','TrainTime(s)': '5010.184030' , 'InfTime_s/Test_Time(s)': '22.830803'},
    {'Task': 'Task-2', 'Model': 'ResNet50',       'Accuracy': '0.58800', 'F1_macro': '0.42 ','TrainTime(s)': '3809.249537 ','InfTime_s/Test_Time(s)': '15.778241'},
    {'Task': 'Task-2', 'Model': 'EfficientNetB0', 'Accuracy': 0.28250, 'F1_macro': 0.13 ,'TrainTime(s)': 3023.964755, 'InfTime_s/Test_Time(s)':14.898023},


        {'Task':'Task-3','Model':'CustomCNN','Accuracy': '0.962894' ,'F1_macro': '0.961255'	, 'TrainTime(s)': '5763.161921' ,'InfTime_s/Test_Time(s)': '9.406214'
}
    ])
new_row = {
    'Task':'Task-4',
    'Model':f'CustomCNN+{ATTENTION_MODULE_NAME}',
    'Accuracy':f'{accuracy:.4f}',
    'F1_macro':f'{f1_macro:.4f}',
    
    'TrainTime(s)': round(training_time_secs,3),
    'InfTime_s/Test_Time(s)': round(inf_mean,3)
}
summary_df = pd.concat([summary_df, pd.DataFrame([new_row])], ignore_index=True)
display(summary_df)
summary_df.to_csv('/kaggle/working/task2_3_4_summary.csv', index=False)
print("Saved concatenated summary to /kaggle/working/task2_3_4_summary.csv")

# ---------------------------
# Save checkpoint & requirements
# ---------------------------
checkpoint = {
    'model_state_dict': model.state_dict(),
    'class_names': new_class_names,
    'params_count': int(att_params),
    'seed': SEED,
    'img_size': IMG_SIZE,
}
torch.save(checkpoint, CHECKPOINT_PATH)
print(f"Saved checkpoint to: {CHECKPOINT_PATH}")

try:
    req_file = '/kaggle/working/requirements_task4.txt'
    os.system(f"{sys.executable} -m pip freeze > {req_file}")
    print(f"Wrote dependencies to: {req_file}" if Path(req_file).exists() else "pip freeze did not create requirements file.")
except Exception as e:
    print("Failed to write requirements:", e)

repro_record = {
    'checkpoint_path': CHECKPOINT_PATH,
    'requirements_file': '/kaggle/working/requirements_task4.txt',
    'dataset': DATASET_NAME,
    'dataset_dir': DATA_DIR,
    'num_classes': NUM_CLASSES,
    'img_size': IMG_SIZE,
    'seed': SEED,
    'train_time_s': training_time_secs,
    'inference_time_s_avg': inf_mean
}
with open('/kaggle/working/task4_repro_record.json','w') as fh:
    json.dump(repro_record, fh, indent=2)
print("Wrote reproducibility record to /kaggle/working/task4_repro_record.json")

print("\nDone — Task 4 (SE attention on tomato subset). Files saved under /kaggle/working:\n- checkpoint (.pth)\n- task2_3_4_summary.csv\n- requirements_task4.txt\n- task4_repro_record.json")
```
